```python
#We must first import the Pandas library. We will shortcut it to "pd".
import pandas as pd

# The following code will process the current filepath for the dataset

#This imports the necessary libraries
import os
from os.path import curdir

'''
The os.path.join function is used to merge strings (text) that correspond to
file and directory names together to a viable file path for your Operating
System. I printed the result to give more clarity.
'''
csv_path = os.path.join(curdir,'Data','MMA_860_Grocery_Data.csv')
print(csv_path)
```

```python
'''
We repeat the above procedure, substituting the excel file name into the
join arguments and using the read_excel function instead.
'''
xl_path = os.path.join(curdir,'Data',"MMA_860_Grocery_Data.csv")
df_xl = pd.read_csv(xl_path,
                    index_col=0)
df_xl.head()
```

```python
df_xl.dtypes
```

```python
#First declare a function that removes a dollar sign from a string.
def remove_dollar_sign(s):
    '''
    The Python String replace function can be used to replace any
    instances of a substring in a string with another substring.
    In this case, we are replacing "$" with "" (an empty string);
    this effectively gets rid of any dollar signs in your string.
    '''
    return s.replace("$","")
```

```python
df_xl["Grocery_Bill"] = df_xl["Grocery_Bill"].apply(remove_dollar_sign)
```

```python
df_xl.head()
```

```python
df_xl["Family_Income"] = df_xl["Family_Income"].apply(remove_dollar_sign)
```

```python
df_xl.head()
```

```python
'''
Family_Income contains a leading dollar sign as well as commas. We must
write an additional function to remove the commas. We can reuse the general
format from the dollar sign.
'''
def remove_comma(s):
    return s.replace(",","")
```

```python
df_xl["Family_Income"] = df_xl["Family_Income"].apply(remove_comma)
```

```python
df_xl.head()
```

```python
df_xl = df_xl.astype({"Family_Income": float, "Grocery_Bill": float})
```

```python
df_xl.dtypes
```

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols
```

```python
model = ols('Grocery_Bill ~ Family_Size + Family_Income + N_Adults + N_Children + N_Vehicles + Distance_to_Store + Vegetarian + Family_Pet',df_xl).fit()
print(model.summary())
```

```python
# Getting rid of vehicles first since bad p value with t stat. Seems not too significant

model = ols('Grocery_Bill ~ Family_Size + Family_Income + N_Adults + N_Children + Distance_to_Store + Vegetarian + Family_Pet',df_xl).fit()
print(model.summary())
```

```python
# Children seems to be bad data as well, however, we will instead remove family size first since we have breakdown of children and adults and they might have different spending habits

model = ols('Grocery_Bill ~  Family_Income + N_Adults + N_Children + Distance_to_Store + Vegetarian + Family_Pet',df_xl).fit()
print(model.summary())
```

```python
# Pet does not influence grocery bill. P value with T stat is different

model = ols('Grocery_Bill ~  Family_Income + N_Adults + N_Children + Distance_to_Store + Vegetarian',df_xl).fit()
print(model.summary())
```
